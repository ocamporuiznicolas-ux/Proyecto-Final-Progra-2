```python
#Vizualiacion de datos faltantes
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.isnull(), cbar=False)
plt.title("Mapa de valores faltantes")
plt.show()
```

```python
nulos = df.isnull().sum()
porcentaje_nulos = (nulos[nulos > 0] / len(df)) * 100

print(porcentaje_nulos)
```

```python
#Manejo de datos faltantes
#Primero medimos cuantos nulos hay por columna y que porcentaje representan del total
nulos = df.isnull().sum()
porcentaje = (nulos / len(df) * 100).round(2)

tabla_nulos = pd.DataFrame({"Nulos": nulos, "Porcentaje (%)": porcentaje})
tabla_nulos = tabla_nulos[tabla_nulos["Nulos"] > 0].sort_values("Porcentaje (%)", ascending=False)
tabla_nulos
```

```python
#Regla 2: eliminar filas cuando los nulos son pocos (<5% del total)
#Todas las columnas estan por debajo del 5%, pero tratamos dos grupos distinto:
#- Columnas con 1 a 3 nulos (0.05% o menos): se eliminan esas filas, son tan pocas que no afectan el analisis
#- Teacher_Quality, Parental_Education_Level y Distance_from_Home (~1%): se imputan (ver celda siguiente)
#  porque si las eliminamos perderiamos 231 estudiantes y casi toda la perdida vendria de solo 3 columnas
columnas_imputar = ["Teacher_Quality", "Parental_Education_Level", "Distance_from_Home"]
columnas_eliminar_filas = [c for c in df.columns if df[c].isnull().sum() > 0 and c not in columnas_imputar]

filas_antes = len(df)
df = df.dropna(subset=columnas_eliminar_filas)
print("Columnas donde se eliminan filas:", columnas_eliminar_filas)
print("Filas eliminadas:", filas_antes - len(df), "(", round((filas_antes - len(df)) / filas_antes * 100, 2), "% )")
```

```python
#Imputacion con la moda en las 3 columnas categoricas
#Justificacion: son variables categoricas (no se puede usar media ni mediana), tienen ~1% de nulos
#y en las tres la categoria mas comun domina claramente (Medium, High School, Near),
#asi que rellenar con la moda casi no cambia la distribucion original
for col in columnas_imputar:
    moda = df[col].mode()[0]
    print(col, "-> se rellena con:", moda, "| nulos:", df[col].isnull().sum())
    df[col] = df[col].fillna(moda)
```

```python
#Vizualiacion de datos faltantes
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.isnull(), cbar=False)
plt.title("Mapa de valores faltantes")
plt.show()
```

SE REALIZARON LAS GRAFICAS CON IA

```python
#Grafico 1: Distribucion del puntaje del examen
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Exam_Score", binwidth=1, kde=True, color="steelblue")
plt.axvline(df["Exam_Score"].mean(), color="red", linestyle="--", label="Promedio")
plt.title("Distribución del Puntaje del Examen")
plt.xlabel("Puntaje del Examen")
plt.ylabel("Cantidad de estudiantes")
plt.legend()
plt.show()

#Conclusion: la mayoria de estudiantes saca entre 60 y 75 (cerca del 98%), con un promedio de 67.2.
#Muy pocos superan 80 (menos del 1%), por eso la distribucion tiene una cola larga hacia la derecha.
```

```python
#Grafico 2: Relacion entre horas de estudio y puntaje del examen
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Hours_Studied", y="Exam_Score", scatter_kws={"alpha": 0.3}, line_kws={"color": "red"})
plt.title("Horas de Estudio vs Puntaje del Examen")
plt.xlabel("Horas de Estudio")
plt.ylabel("Puntaje del Examen")
plt.show()

print("Correlacion:", round(df["Hours_Studied"].corr(df["Exam_Score"]), 2))

#Conclusion: a mas horas de estudio, mayor puntaje (correlacion positiva moderada de 0.45).
#Quienes estudian 10 horas o menos promedian cerca de 64, y quienes estudian mas de 30 horas promedian cerca de 71.
```

```python
#Grafico 3: Relacion entre asistencia y puntaje del examen
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Attendance", y="Exam_Score", scatter_kws={"alpha": 0.3}, line_kws={"color": "green"})
plt.title("Asistencia vs Puntaje del Examen")
plt.xlabel("Asistencia (%)")
plt.ylabel("Puntaje del Examen")
plt.show()

print("Correlacion:", round(df["Attendance"].corr(df["Exam_Score"]), 2))

#Conclusion: la asistencia es la variable que mas se relaciona con el puntaje (correlacion de 0.58).
#Los estudiantes con 60-70% de asistencia promedian 64.3, mientras que los de 90-100% promedian 70.1.
```

```python
#Grafico 4: Puntaje promedio segun involucramiento de los padres y acceso a recursos
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="Parental_Involvement", y="Exam_Score", hue="Access_to_Resources",
            order=["Low", "Medium", "High"], hue_order=["Low", "Medium", "High"], palette="Set2")
plt.ylim(60, 72)
plt.title("Puntaje Promedio por Involucramiento Parental y Acceso a Recursos")
plt.xlabel("Involucramiento de los Padres")
plt.ylabel("Puntaje Promedio")
plt.legend(title="Acceso a Recursos")
plt.show()

#Conclusion: a mayor involucramiento de los padres y mayor acceso a recursos, mayor puntaje.
#Los estudiantes con ambos en nivel bajo promedian 65.2, y con ambos en nivel alto promedian 69.1 (casi 4 puntos mas).
```

```python
#Grafico 5: Mapa de calor de correlaciones entre variables numericas
columnas_numericas = ["Hours_Studied", "Attendance", "Sleep_Hours", "Previous_Scores",
                      "Tutoring_Sessions", "Physical_Activity", "Exam_Score"]

plt.figure(figsize=(9, 7))
sns.heatmap(df[columnas_numericas].corr(), annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Mapa de Calor de Correlaciones")
plt.show()

#Conclusion: las variables que mas influyen en el puntaje son la asistencia (0.58) y las horas de estudio (0.45).
#Puntajes previos (0.18) y tutorias (0.16) influyen poco, y las horas de sueño (-0.02) y la actividad fisica (0.03)
#practicamente no tienen relacion con el resultado del examen.
```
